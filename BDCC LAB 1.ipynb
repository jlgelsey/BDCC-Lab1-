```python
import pandas as pd
data = "DisneylandReviews.csv"
df = pd.read_csv(data, encoding="latin-1")
df.head()
```

```python
df.shape
```

```python
# Drop rows where the review text might be missing
text_data = df['Review_Text'].dropna().astype(str)

print(f"Total reviews to process: {len(text_data)}")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Initialize the TF-IDF Vectorizer
# max_features limits the vocabulary size, stop_words removes common English words
tfidf = TfidfVectorizer(max_features=5000, stop_words='english')

# Fit and transform the review text data
tfidf_matrix = tfidf.fit_transform(text_data)

print("TF-IDF Matrix shape:", tfidf_matrix.shape)
```

```python
# Get feature names (words)
feature_names = tfidf.get_feature_names_out()

# Create a small DataFrame to inspect TF-IDF scores for the first few reviews
sample_df = pd.DataFrame(tfidf_matrix[:5].toarray(), columns=feature_names)

# Display non-zero terms for the first review
print("Top terms in the first review:")
print(sample_df.loc[0][sample_df.loc[0] > 0].sort_values(ascending=False).head(10))
```

```python
# 3. Initialize and fit the TF-IDF Vectorizer
vectorizer = TfidfVectorizer(max_features=5000, stop_words='english')
tfidf_matrix = vectorizer.fit_transform(text_data)

# 4. Compute Exploratory Data Analysis (EDA) metrics
sums = tfidf_matrix.sum(axis=0).A1
feature_names = vectorizer.get_feature_names_out()
top_words = sorted(zip(feature_names, sums), key=lambda x: x[1], reverse=True)[:10]

sparsity = (1.0 - tfidf_matrix.nnz / (tfidf_matrix.shape[0] * tfidf_matrix.shape[1])) * 100

# 5. Print out the EDA summary results
print(f"Total Reviews Analyzed: {len(text_data):,}")
print(f"TF-IDF Matrix Shape: {tfidf_matrix.shape}")
print(f"Matrix Sparsity: {sparsity:.2f}%\n")

print("Top 10 Words by Total TF-IDF Score:")
print(f"{'Rank':<6} | {'Word':<15} | {'Total TF-IDF Score'}")
print("-" * 42)
for rank, (word, score) in enumerate(top_words, 1):
    print(f"{rank:<6} | {word:<15} | {score:.2f}")
```
